```python
# Simulation: SC vs AFM excitations with material+stack presets (starting with Indium Oxide)
# This script generates:
# 1) A materials library you can extend (incl. InOx superconducting film)
# 2) BCS DOS (singlet), triplet DOS (unitary/non-unitary), AFM magnon dispersion
# 3) Example plots + CSV export for Indium Oxide
#
# You can rerun cells after editing parameters to explore other materials/stacks.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from math import pi

# ---------- Constants ----------
k_B_eV_per_K = 8.617333262145e-5  # eV/K

# ---------- Helper functions ----------
def dynes_dos(E, Delta, gamma):
    """
    BCS/Dynes DOS: N(E)/N0 = Re[(E + i*gamma) / sqrt((E + i*gamma)^2 - Delta^2)]
    Works for both singlet BCS and fully-gapped unitary triplet (same DOS shape).
    """
    z = E + 1j*gamma
    return np.real((z) / np.sqrt(z**2 - Delta**2))

def zeeman_split_dos(E, Delta_up, Delta_down, h, gamma):
    """
    Crude spin-resolved DOS model for spin-polarized/non-unitary triplet or SC under exchange field h.
    N_up(E) ~ Re[(E+h+i*gamma)/sqrt((E+h+i*gamma)^2 - Delta_up^2)]
    N_dn(E) ~ Re[(E-h+i*gamma)/sqrt((E-h+i*gamma)^2 - Delta_down^2)]
    Returns total DOS (sum) and components.
    """
    z_up = E + h + 1j*gamma
    z_dn = E - h + 1j*gamma
    N_up = np.real(z_up / np.sqrt(z_up**2 - Delta_up**2))
    N_dn = np.real(z_dn / np.sqrt(z_dn**2 - Delta_down**2))
    return N_up + N_dn, N_up, N_dn

def tb2d_dispersion(kx, ky, t=1.0, mu=0.0):
    """2D nearest-neighbor tight-binding dispersion: epsilon_k = -2t(cos kx + cos ky) - mu"""
    return -2*t*(np.cos(kx) + np.cos(ky)) - mu

def bogoliubov_spectrum_along_path(Delta, t=1.0, mu=0.0, npts=200):
    """
    Compute |E_k| = sqrt(eps_k^2 + Delta^2) along a high-symmetry path in 2D:
    Γ(0,0) -> X(pi,0) -> M(pi,pi) -> Γ(0,0)
    Returns: k_path (1D param coord), E_plus (quasiparticle), eps (normal-state)
    """
    path = [(0,0),(pi,0),(pi,pi),(0,0)]
    segs = []
    for (ax,ay),(bx,by) in zip(path[:-1], path[1:]):
        kx = np.linspace(ax, bx, npts, endpoint=False)
        ky = np.linspace(ay, by, npts, endpoint=False)
        segs.append((kx,ky))
    # concatenate with the final point
    kx = np.concatenate([s[0] for s in segs] + [np.array([path[-1][0]])])
    ky = np.concatenate([s[1] for s in segs] + [np.array([path[-1][1]])])

    eps = tb2d_dispersion(kx, ky, t=t, mu=mu)
    Eqp = np.sqrt(eps**2 + Delta**2)
    k_param = np.arange(len(kx))  # parametric coordinate
    return k_param, Eqp, eps

def afm_magnon_square_lattice(qx, qy, J_meV=10.0, S=1.0):
    """
    Linear spin-wave dispersion for a 2D square-lattice AFM:
    ω(q) = 2 J S sqrt(1 - γ_q^2), with γ_q = (cos qx + cos qy)/2
    Units: meV if J_meV provided in meV
    """
    gamma_q = 0.5*(np.cos(qx) + np.cos(qy))
    w = 2.0 * J_meV * S * np.sqrt(1.0 - gamma_q**2)
    return w

# ---------- Materials library (extendable) ----------
# Note: These are illustrative "toy" parameters geared for qualitative exploration.
materials = {
    # Indium Oxide (amorphous InOx thin film): disordered SC with tunable Tc.
    # We'll assume Tc ≈ 3 K as a starting point; Delta0 = 1.76 k_B Tc.
    "InOx_SC": {
        "kind": "SC_singlet",
        "Tc_K": 3.0,
        "Delta0_eV": 1.76 * k_B_eV_per_K * 3.0,
        "dynes_gamma_frac": 0.02,  # gamma = frac * Delta0
        "tb_t_eV": 0.25,  # illustrative TB kinetic energy scale
        "tb_mu_eV": 0.0
    },
    # Example AFM insulator (like NiO, illustrative values)
    "NiO_AFM": {
        "kind": "AFM",
        "J_meV": 12.0,
        "S": 1.0
    },
    # Example ferromagnet (metal) for FM+SC stacks
    "Fe_FM": {
        "kind": "FM",
        "exchange_h_eV": 0.06  # small exchange splitting (toy value)
    },
    # Example triplet SC (unitary, fully gapped); DOS similar to BCS but label differs
    "Triplet_Unitary_SC": {
        "kind": "SC_triplet_unitary",
        "Tc_K": 1.5,
        "Delta0_eV": 1.76 * k_B_eV_per_K * 1.5,
        "dynes_gamma_frac": 0.02,
        "tb_t_eV": 0.25,
        "tb_mu_eV": 0.0
    },
    # Example non-unitary (spin-polarized) triplet SC
    "Triplet_NonUnitary_SC": {
        "kind": "SC_triplet_nonunitary",
        "Tc_K": 1.0,
        "Delta_up_eV": 1.76 * k_B_eV_per_K * 1.0,
        "Delta_dn_eV": 0.6 * (1.76 * k_B_eV_per_K * 1.0),  # smaller gap on one spin
        "exchange_h_eV": 0.02,
        "dynes_gamma_frac": 0.02
    },
}

# Display materials library
#from caas_jupyter_tools import display_dataframe_to_user
#materials_df = pd.DataFrame(materials).T
#display_dataframe_to_user("Materials Library (editable by running this cell after editing dict)", materials_df)

# ---------- Simulation: Indium Oxide (InOx) ----------
# 1) DOS
InOx = materials["InOx_SC"]
Delta = InOx["Delta0_eV"]
gamma = InOx["dynes_gamma_frac"] * Delta

E = np.linspace(-5*Delta, 5*Delta, 2000)
N = dynes_dos(E, Delta, gamma)

# Save CSV
dos_df = pd.DataFrame({"E_eV": E, "N_over_N0": N})
dos_path = "/content/dos/dos_InOx.csv"
dos_df.to_csv(dos_path, index=False)

# Plot DOS
plt.figure(figsize=(7,4.5))
plt.plot(E, N)
plt.axvline(-Delta, linestyle="--")
plt.axvline(Delta, linestyle="--")
plt.title("InOx (singlet BCS/Dynes) DOS")
plt.xlabel("Energy E (eV)")
plt.ylabel("N(E)/N0")
plt.tight_layout()
plt.show()

# 2) Bogoliubov dispersion along Γ–X–M–Γ (toy 2D TB band)
k_param, Eqp, eps = bogoliubov_spectrum_along_path(
    Delta=Delta, t=InOx["tb_t_eV"], mu=InOx["tb_mu_eV"], npts=200
)

plt.figure(figsize=(7,4.5))
plt.plot(k_param, eps, label="Normal ε_k")
plt.plot(k_param, Eqp, label="+E_k = sqrt(ε_k^2 + Δ^2)")
plt.plot(k_param, -Eqp, label="-E_k")
plt.title("InOx Toy Bogoliubov Spectrum (2D TB)")
plt.xlabel("k-path index (Γ–X–M–Γ)")
plt.ylabel("Energy (eV)")
plt.legend()
plt.tight_layout()
plt.show()

# 3) AFM magnon dispersion (example grid for NiO-like AFM)
# Even though the request starts with InOx, we include this to enable stack comparisons.
qpts = 201
qx = np.linspace(-pi, pi, qpts)
qy = np.linspace(-pi, pi, qpts)
QX, QY = np.meshgrid(qx, qy, indexing="ij")
NiO = materials["NiO_AFM"]
omega_meV = afm_magnon_square_lattice(QX, QY, J_meV=NiO["J_meV"], S=NiO["S"])

# Show a cut along qy=0 for clarity
mid = qpts//2
plt.figure(figsize=(7,4.5))
plt.plot(qx, omega_meV[:, mid])
plt.title("AFM Magnon Dispersion (Square Lattice, cut q_y=0)")
plt.xlabel("q_x (rad)")
plt.ylabel("ℏω (meV)")
plt.tight_layout()
plt.show()

# 4) Example: non-unitary triplet DOS (spin-polarized) – not InOx, but ready for stacks
trip = materials["Triplet_NonUnitary_SC"]
E_trip = np.linspace(-5*trip["Delta_up_eV"], 5*trip["Delta_up_eV"], 2000)
gamma_trip = trip["dynes_gamma_frac"] * trip["Delta_up_eV"]
Ntot, Nup, Ndn = zeeman_split_dos(E_trip, trip["Delta_up_eV"], trip["Delta_dn_eV"], trip["exchange_h_eV"], gamma_trip)

plt.figure(figsize=(7,4.5))
plt.plot(E_trip, Ntot, label="Total")
plt.plot(E_trip, Nup, label="↑")
plt.plot(E_trip, Ndn, label="↓")
plt.title("Non-Unitary Triplet SC: Spin-Resolved DOS (illustrative)")
plt.xlabel("Energy (eV)")
plt.ylabel("N(E)/N0 (arb.)")
plt.legend()
plt.tight_layout()
plt.show()

# 5) Utility to evaluate a "stack"
def simulate_stack(layers, energy_window_eV=None, points=2000):
    """
    layers: list of material keys (strings) referencing 'materials' dict.
    Returns dictionary of spectra per layer, using simple illustrative models:
      - SC_singlet or SC_triplet_unitary: BCS/Dynes DOS
      - SC_triplet_nonunitary: spin-resolved DOS
      - AFM: magnon cut along qy=0
      - FM: Stoner-like DOS not implemented; shows only exchange splitting value
    If energy_window_eV is None, chooses based on gaps present.
    """
    result = {"layers": layers, "spectra": {}}
    # choose energy window
    gaps = []
    for key in layers:
        kind = materials[key]["kind"]
        if kind in ("SC_singlet", "SC_triplet_unitary"):
            gaps.append(materials[key]["Delta0_eV"])
        elif kind == "SC_triplet_nonunitary":
            gaps.append(max(materials[key]["Delta_up_eV"], materials[key]["Delta_dn_eV"]))
    if energy_window_eV is None:
        Emax = 5*(max(gaps) if gaps else 0.01)
    else:
        Emax = energy_window_eV/2.0
    E = np.linspace(-Emax, Emax, points)

    for key in layers:
        info = materials[key]
        kind = info["kind"]
        if kind in ("SC_singlet", "SC_triplet_unitary"):
            Delta = info["Delta0_eV"]
            gamma = info["dynes_gamma_frac"] * Delta
            N = dynes_dos(E, Delta, gamma)
            result["spectra"][key] = {"E_eV": E, "N_over_N0": N, "type": kind}
        elif kind == "SC_triplet_nonunitary":
            Delta_up = info["Delta_up_eV"]
            Delta_dn = info["Delta_dn_eV"]
            h = info["exchange_h_eV"]
            gamma = info["dynes_gamma_frac"] * max(Delta_up, Delta_dn)
            Ntot, Nup, Ndn = zeeman_split_dos(E, Delta_up, Delta_dn, h, gamma)
            result["spectra"][key] = {"E_eV": E, "N_total": Ntot, "N_up": Nup, "N_dn": Ndn, "type": kind}
        elif kind == "AFM":
            qx = np.linspace(-pi, pi, 201)
            qy = np.linspace(-pi, pi, 201)
            QX, QY = np.meshgrid(qx, qy, indexing="ij")
            w = afm_magnon_square_lattice(QX, QY, J_meV=info["J_meV"], S=info["S"])
            result["spectra"][key] = {"qx": qx, "w_cut_meV": w[:, qy.size//2], "type": kind}
        elif kind == "FM":
            result["spectra"][key] = {"exchange_h_eV": info["exchange_h_eV"], "type": kind}
        else:
            result["spectra"][key] = {"note": "unknown kind"}
    return result

# Demo stack: InOx_SC only
stack_result = simulate_stack(["InOx_SC"])

# Export stack spectra to CSV files
export_paths = []
for key, spec in stack_result["spectra"].items():
    if spec["type"] in ("SC_singlet", "SC_triplet_unitary"):
        df = pd.DataFrame({"E_eV": spec["E_eV"], "N_over_N0": spec["N_over_N0"]})
        path = f"/content/dos/stack_{key}_DOS.csv"
        df.to_csv(path, index=False)
        export_paths.append(path)
    elif spec["type"] == "SC_triplet_nonunitary":
        df = pd.DataFrame({"E_eV": spec["E_eV"], "N_total": spec["N_total"], "N_up": spec["N_up"], "N_dn": spec["N_dn"]})
        path = f"/content/stack_{key}_TripletDOS.csv"
        df.to_csv(path, index=False)
        export_paths.append(path)
    elif spec["type"] == "AFM":
        df = pd.DataFrame({"qx_rad": spec["qx"], "omega_meV": spec["w_cut_meV"]})
        path = f"/content/stack_{key}_MagnonCut.csv"
        df.to_csv(path, index=False)
        export_paths.append(path)
    elif spec["type"] == "FM":
        df = pd.DataFrame({"exchange_h_eV": [spec["exchange_h_eV"]]})
        path = f"/content/stack_{key}_FMParams.csv"
        df.to_csv(path, index=False)
        export_paths.append(path)

# Print exported paths for convenience
export_paths
```

```python
# Add: stack plotter and a simple SC/FM proximity toy (phenomenological).
# This cell builds on the previously-defined helpers and materials dict.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from math import pi

# ---- Reuse/define helper DOS functions (positive-defined) ----
def dynes_dos_fixed(E, Delta, gamma):
    z = E + 1j*gamma
    val = z / np.sqrt(z**2 - Delta**2)
    return np.abs(np.real(val))

def zeeman_split_dos_fixed(E, Delta_up, Delta_down, h, gamma):
    z_up = E + h + 1j*gamma
    z_dn = E - h + 1j*gamma
    N_up = np.abs(np.real(z_up / np.sqrt(z_up**2 - Delta_up**2)))
    N_dn = np.abs(np.real(z_dn / np.sqrt(z_dn**2 - Delta_down**2)))
    return N_up + N_dn, N_up, N_dn

# ---- Stack plotter ----
def plot_stack_spectra(stack_result, plot_triplet_components=False, title_prefix="Stack"):
    """
    Plots spectra for each layer in the stack_result returned by simulate_stack().
    Creates separate figures per spectral type (DOS vs E; magnon cut vs q; FM params).
    """
    for key, spec in stack_result["spectra"].items():
        typ = spec.get("type", "unknown")
        if typ in ("SC_singlet", "SC_triplet_unitary"):
            E = spec["E_eV"]; N = spec["N_over_N0"]
            plt.figure(figsize=(7,4.5))
            plt.plot(E, N)
            plt.title(f"{title_prefix}: {key} DOS")
            plt.xlabel("Energy (eV)"); plt.ylabel("N(E)/N0")
            plt.tight_layout(); plt.show()
        elif typ == "SC_triplet_nonunitary":
            E = spec["E_eV"]; Ntot = spec["N_total"]
            plt.figure(figsize=(7,4.5))
            plt.plot(E, Ntot, label="Total")
            if plot_triplet_components:
                plt.plot(E, spec["N_up"], label="↑")
                plt.plot(E, spec["N_dn"], label="↓")
                plt.legend()
            plt.title(f"{title_prefix}: {key} Triplet DOS")
            plt.xlabel("Energy (eV)"); plt.ylabel("N(E)/N0 (arb.)")
            plt.tight_layout(); plt.show()
        elif typ == "AFM":
            qx = spec["qx"]; w = spec["w_cut_meV"]
            plt.figure(figsize=(7,4.5))
            plt.plot(qx, w)
            plt.title(f"{title_prefix}: {key} AFM Magnon (q_y=0 cut)")
            plt.xlabel("q_x (rad)"); plt.ylabel("ℏω (meV)")
            plt.tight_layout(); plt.show()
        elif typ == "FM":
            # Visualize exchange parameter as a simple horizontal marker
            h = spec["exchange_h_eV"]
            plt.figure(figsize=(7,2.5))
            plt.plot([0,1],[h,h])
            plt.title(f"{title_prefix}: {key} FM exchange (toy)")
            plt.xlabel("index"); plt.ylabel("h (eV)")
            plt.tight_layout(); plt.show()

# ---- SC/FM proximity (toy) ----
def proximity_sc_fm(SC, FM, tau=0.5, dF=2.0, xiF=3.0, dS=2.0, xiS=10.0, points=2000):
    """
    Very simple phenomenological proximity model (for intuition only):
      - In FM layer, induced superconductivity weight: wF = tau * exp(-dF/xiF).
        DOS_F(E) = (1 - wF)*1  + wF * N_S_split(E; Δ_ind, h_ind, γ_ind)
      - In S near interface, pair-breaking & spin-splitting leaking from FM:
        wS = tau * exp(-dS/xiS)
        Δ_S_eff = Δ_S * (1 - 0.5*wS)
        h_S = h_FM * 0.3 * wS
        γ_S_eff = γ_S * (1 + 2*wS)
      - Spin-split DOS uses Δ↑=Δ↓ for simplicity (unitary split), but shifted by ±h.
    Inputs:
      SC: dict with keys Delta0_eV, dynes_gamma_frac
      FM: dict with key exchange_h_eV
    Returns dict with E, DOS_S, DOS_F, and used parameters.
    """
    Delta_S = SC["Delta0_eV"]
    gamma_S = SC["dynes_gamma_frac"] * Delta_S
    h_FM = FM["exchange_h_eV"]

    # Coupling weights
    wF = tau * np.exp(-dF/xiF)
    wS = tau * np.exp(-dS/xiS)

    # Effective parameters on S side
    Delta_S_eff = Delta_S * (1 - 0.5*wS)
    h_S = h_FM * 0.3 * wS
    gamma_S_eff = gamma_S * (1 + 2*wS)

    # Energy window around the largest scale
    Emax = 5*max(Delta_S_eff, Delta_S)
    E = np.linspace(-Emax, Emax, points)

    # S-side DOS with small spin splitting (treat as equal Δ but Zeeman shift)
    Ntot_S, Nup_S, Ndn_S = zeeman_split_dos_fixed(E, Delta_S_eff, Delta_S_eff, h_S, gamma_S_eff)

    # FM side induced minigap parameters
    Delta_ind = Delta_S * wF
    h_ind = h_FM * (1 - 0.5*wF)
    gamma_ind = gamma_S * (1 + wF)

    if Delta_ind > 0:
        Ntot_F, Nup_F, Ndn_F = zeeman_split_dos_fixed(E, Delta_ind, Delta_ind, h_ind, gamma_ind)
        # Mix with normal DOS=1
        DOS_F = (1 - wF)*np.ones_like(E) + wF*Ntot_F
    else:
        DOS_F = np.ones_like(E)

    out = {
        "E_eV": E,
        "DOS_S_total": Ntot_S,
        "DOS_S_up": Nup_S,
        "DOS_S_dn": Ndn_S,
        "DOS_F_total": DOS_F,
        "params": {
            "wF": float(wF), "wS": float(wS),
            "Delta_S_eff_eV": float(Delta_S_eff),
            "h_S_eV": float(h_S),
            "gamma_S_eff_eV": float(gamma_S_eff),
            "Delta_ind_eV": float(Delta_ind),
            "h_ind_eV": float(h_ind),
            "gamma_ind_eV": float(gamma_ind),
        }
    }
    return out

# ---- Demo: plotter + proximity for InOx_SC | Fe_FM ----

# Regenerate a stack combining InOx and Fe (simulate_stack must exist from previous cell)
try:
    stack_InOx_Fe = simulate_stack(["InOx_SC","Fe_FM","NiO_AFM"])
except NameError:
    # Recreate minimal simulate_stack with materials from prior cell if needed
    def tb2d_dispersion(kx, ky, t=1.0, mu=0.0):
        return -2*t*(np.cos(kx) + np.cos(ky)) - mu
    def afm_magnon_square_lattice(qx, qy, J_meV=10.0, S=1.0):
        gamma_q = 0.5*(np.cos(qx) + np.cos(qy))
        return 2.0 * J_meV * S * np.sqrt(1.0 - gamma_q**2)
    def simulate_stack(layers, energy_window_eV=None, points=2000):
        result = {"layers": layers, "spectra": {}}
        gaps = []
        for key in layers:
            kind = materials[key]["kind"]
            if kind in ("SC_singlet", "SC_triplet_unitary"):
                gaps.append(materials[key]["Delta0_eV"])
            elif kind == "SC_triplet_nonunitary":
                gaps.append(max(materials[key]["Delta_up_eV"], materials[key]["Delta_dn_eV"]))
        if energy_window_eV is None:
            Emax = 5*(max(gaps) if gaps else 0.01)
        else:
            Emax = energy_window_eV/2.0
        E = np.linspace(-Emax, Emax, points)
        for key in layers:
            info = materials[key]; kind = info["kind"]
            if kind in ("SC_singlet", "SC_triplet_unitary"):
                Delta = info["Delta0_eV"]; gamma = info["dynes_gamma_frac"] * Delta
                N = dynes_dos_fixed(E, Delta, gamma)
                result["spectra"][key] = {"E_eV": E, "N_over_N0": N, "type": kind}
            elif kind == "SC_triplet_nonunitary":
                Delta_up = info["Delta_up_eV"]; Delta_dn = info["Delta_dn_eV"]
                h = info["exchange_h_eV"]; gamma = info["dynes_gamma_frac"] * max(Delta_up, Delta_dn)
                Ntot, Nup, Ndn = zeeman_split_dos_fixed(E, Delta_up, Delta_dn, h, gamma)
                result["spectra"][key] = {"E_eV": E, "N_total": Ntot, "N_up": Nup, "N_dn": Ndn, "type": kind}
            elif kind == "AFM":
                qx = np.linspace(-pi, pi, 201); qy = np.linspace(-pi, pi, 201)
                QX, QY = np.meshgrid(qx, qy, indexing="ij")
                w = afm_magnon_square_lattice(QX, QY, J_meV=info["J_meV"], S=info["S"])
                result["spectra"][key] = {"qx": qx, "w_cut_meV": w[:, qy.size//2], "type": kind}
            elif kind == "FM":
                result["spectra"][key] = {"exchange_h_eV": info["exchange_h_eV"], "type": kind}
        return result
    stack_InOx_Fe = simulate_stack(["InOx_SC","Fe_FM","NiO_AFM"])

# Plot the stack
plot_stack_spectra(stack_InOx_Fe, plot_triplet_components=False, title_prefix="InOx | Fe | NiO")

# Proximity calculation (toy) for InOx_SC (S) next to Fe_FM (F)
SC = materials["InOx_SC"]
FM = materials["Fe_FM"]
prox = proximity_sc_fm(SC, FM, tau=0.6, dF=2.0, xiF=3.0, dS=2.0, xiS=10.0, points=2000)

# Plot S-side DOS (total, with slight spin split)
plt.figure(figsize=(7,4.5))
plt.plot(prox["E_eV"], prox["DOS_S_total"])
plt.title("S-side DOS near S|F interface (toy)")
plt.xlabel("Energy (eV)"); plt.ylabel("N(E)/N0")
plt.tight_layout(); plt.show()

# Plot FM-side induced DOS
plt.figure(figsize=(7,4.5))
plt.plot(prox["E_eV"], prox["DOS_F_total"])
plt.title("F-side induced DOS (toy)")
plt.xlabel("Energy (eV)"); plt.ylabel("N(E)/N0")
plt.tight_layout(); plt.show()

# Export proximity spectra and parameters
df_S = pd.DataFrame({"E_eV": prox["E_eV"], "DOS_S_total": prox["DOS_S_total"]})
df_F = pd.DataFrame({"E_eV": prox["E_eV"], "DOS_F_total": prox["DOS_F_total"]})
params_df = pd.DataFrame([prox["params"]])

path_S = "/content/dos/proximity_S_DOS.csv"
path_F = "/content/dos/proximity_F_DOS.csv"
path_P = "/content/dos/proximity_params.json"
df_S.to_csv(path_S, index=False)
df_F.to_csv(path_F, index=False)
params_df.to_json(path_P, orient="records", indent=2)

[path_S, path_F, path_P]
```

```python
# Odd-frequency pairing toy for InOx under magnetic field
# Goal: show field-induced odd-ω weight -> gapless DOS at E=0 (can mimic "metallic" look)
# while singlet Δ_even is suppressed; at high fields → normal state.
#
# Phenomenology:
#   Δ_eff(E,B) = Δ_e(B) + i * η(B) * E,   η(B) ≥ 0 grows with field & disorder (spin-orbit)
#   DOS N/N0   = Re[(E+iΓ_tot(B)) / sqrt((E+iΓ_tot(B))^2 - Δ_eff(E,B)^2)]
#
# where
#   Δ_e(B)      = Δ0 * sqrt(max(0, 1 - (B/Bc2)^2))
#   Γ_tot(B)    = γ0*Δ0 + Γ_pb0 * (B/Bc2)^2  (Dynes broadening + pair-breaking)
#   η(B)        = η0 * (B/Bc2)^p
# Optional Zeeman splitting (equal-spin mixing): small 'h(B)' used to split DOS if desired.
#
# This is a *toy* quasiclassical-like model for intuition, not a substitute for full Usadel/Eliashberg.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ---- Constants ----
mu_B_eV_T = 5.7883818060e-5  # eV/T
g_eff = 2.0

# ---- InOx baseline parameters (editable) ----
InOx_params = dict(
    Tc_K = 3.0,
    Delta0_eV = 1.76 * 8.617333262145e-5 * 3.0,  # ~0.000455 eV
    Bc2_T = 10.0,          # illustrative upper critical field scale
    gamma_dynes_frac = 0.02, # baseline Dynes
    Gamma_pb0_frac = 0.03, # pair-breaking strength scale (fraction of Δ0 at B=Bc2)
    eta0 = 0.8,            # odd-frequency strength at B=Bc2 (dimensionless)
    eta_power = 1.2,       # field exponent for η(B)
    include_zeeman = True,
    zeeman_scale = 0.25    # scale < 1 to reduce visible splitting under strong disorder
)

def Delta_even(B, p):
    Δ0 = p["Delta0_eV"]
    Bc2 = p["Bc2_T"]
    x = max(0.0, 1.0 - (B/Bc2)**2)
    return Δ0 * np.sqrt(x)

def eta_odd(B, p):
    return p["eta0"] * (B/p["Bc2_T"])**p["eta_power"]

def Gamma_tot(B, p):
    Δ0 = p["Delta0_eV"]
    γ0 = p["gamma_dynes_frac"] * Δ0
    Γpb = p["Gamma_pb0_frac"] * Δ0 * (B/p["Bc2_T"])**2
    return γ0 + Γpb

def zeeman_h(B, p):
    if not p["include_zeeman"]:
        return 0.0
    return p["zeeman_scale"] * (g_eff * mu_B_eV_T * B / 2.0)

def oddfreq_dos(E, B, p):
    Δe = Delta_even(B, p)
    η  = eta_odd(B, p)
    Γ  = Gamma_tot(B, p)
    h  = zeeman_h(B, p)

    # Energy-dependent complex order parameter (even + i * odd * E)
    # For spin splitting, evaluate for ↑ and ↓ with E -> E±h, then average.
    z_up = (E + h) + 1j*Γ
    z_dn = (E - h) + 1j*Γ
    Δeff_up = Δe + 1j*η*(E + h)
    Δeff_dn = Δe + 1j*η*(E - h)

    Nu = np.abs(np.real(z_up / np.sqrt(z_up**2 - Δeff_up**2)))
    Nd = np.abs(np.real(z_dn / np.sqrt(z_dn**2 - Δeff_dn**2)))
    N  = 0.5*(Nu + Nd)
    return N, Nu, Nd, dict(Delta_even=Δe, eta=η, Gamma=Γ, h=h)

# ---- Sweep fields and plot overlays ----
def plot_odd_freq_overlays(B_list, p=InOx_params, Emin=-0.004, Emax=0.004, nE=2500, prefix="InOx_odd"):
    E = np.linspace(Emin, Emax, nE)
    summary = []

    plt.figure(figsize=(7,4.5))
    for B in B_list:
        N, Nu, Nd, meta = oddfreq_dos(E, B, p)
        plt.plot(E, N, label=f"B={B:.1f} T")
        N0 = float(np.interp(0.0, E, N))
        summary.append(dict(B_T=B, N0=N0, **meta))
        # export per-field DOS
        df = pd.DataFrame({"E_eV": E, "N_over_N0": N, "N_up": Nu, "N_dn": Nd})
        df.to_csv(f"/content/dos/{prefix}_DOS_B{B:.1f}T.csv", index=False)
    plt.title("InOx DOS with odd-ω mixing vs field (toy)")
    plt.xlabel("Energy (eV)"); plt.ylabel("N(E)/N0")
    plt.legend(); plt.tight_layout(); plt.show()

    # Zero-energy DOS vs B (proxy for 'metallic-like' behavior)
    summ_df = pd.DataFrame(summary)
    summ_df.to_csv(f"/content/dos/{prefix}_summary.csv", index=False)

    plt.figure(figsize=(7,4.5))
    plt.plot(summ_df["B_T"], summ_df["N0"], marker="o")
    plt.axvline(InOx_params["Bc2_T"], linestyle="--")
    plt.title("Zero-energy DOS N(0)/N0 vs Magnetic Field")
    plt.xlabel("B (T)"); plt.ylabel("N(0)/N0")
    plt.tight_layout(); plt.show()

    return E, summ_df

# ---- Run an illustrative field sweep ----
B_sweep = [0.0, 2.5, 5.0, 7.5, 9.5, 12.0]  # crosses Bc2=10 T
E_axis, summary_df = plot_odd_freq_overlays(B_sweep, p=InOx_params, Emin=-0.004, Emax=0.004, nE=2500, prefix="InOx_odd")

# ---- Very rough "sheet resistance" proxy from N(0) ----
# Proxy: R(B)/Rn ≈ 1 / (ε + N(0)/N0), with small ε to avoid divergence at low fields.
# This mimics that more quasiparticles (gapless) → lower resistance, while at large B → R → 1.
eps_floor = 0.02
Rnorm = 1.0 / (eps_floor + summary_df["N0"].values)
Rnorm = Rnorm / Rnorm[-1]  # normalize to 1 at highest field (approx normal state)
plt.figure(figsize=(7,4.5))
plt.plot(summary_df["B_T"], Rnorm, marker="s")
plt.title("Toy sheet resistance R/Rn vs B (from N(0))")
plt.xlabel("B (T)"); plt.ylabel("R/Rn (toy)")
plt.tight_layout(); plt.show()

# Export the R(B) proxy
rb_path = "/content/dos/InOx_odd_R_over_Rn_toy.csv"
pd.DataFrame({"B_T": summary_df["B_T"], "R_over_Rn_toy": Rnorm}).to_csv(rb_path, index=False)

# Return file paths for convenience
[f"/content/dos/InOx_odd_DOS_B{B:.1f}T.csv" for B in B_sweep] + ["/content/dos/InOx_odd_summary.csv", rb_path]
```

```python
# Optional: emergent FM window in InOx odd-ω model (for hypothesis testing)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def h_emergent_FM(B, fm_window):
    """
    fm_window: (B1, B2, h_max_eV), triangular window peaking at mid (B1+B2)/2.
    Returns extra exchange splitting added to Zeeman.
    """
    if fm_window is None:
        return 0.0
    B1, B2, hmax = fm_window
    if B <= B1 or B >= B2 or B2 <= B1:
        return 0.0
    mid = 0.5*(B1+B2)
    width = (B2-B1)/2.0
    # Triangular profile rising from B1 to mid and falling to B2
    return hmax * max(0.0, 1.0 - abs(B - mid)/width)

def oddfreq_dos_with_fm(E, B, p, fm_window=None):
    # same as oddfreq_dos but adds emergent FM exchange in a specified field window
    Δe = Delta_even(B, p)
    η  = eta_odd(B, p)
    Γ  = Gamma_tot(B, p)
    hZ = zeeman_h(B, p)
    hF = h_emergent_FM(B, fm_window)
    h  = hZ + hF

    z_up = (E + h) + 1j*Γ
    z_dn = (E - h) + 1j*Γ
    Δeff_up = Δe + 1j*η*(E + h)
    Δeff_dn = Δe + 1j*η*(E - h)

    Nu = np.abs(np.real(z_up / np.sqrt(z_up**2 - Δeff_up**2)))
    Nd = np.abs(np.real(z_dn / np.sqrt(z_dn**2 - Δeff_dn**2)))
    N  = 0.5*(Nu + Nd)
    return N, Nu, Nd, dict(Delta_even=Δe, eta=η, Gamma=Γ, h_total=h, h_zeeman=hZ, h_emergent=hF)

# Demo: try a narrow FM window just below Bc2
B_list = [9.0, 9.5, 10.0]
E = np.linspace(-0.004, 0.004, 2500)
fm_window = (9.2, 9.9, 1.5e-4)  # peaks at ~9.55 T with extra 0.15 meV exchange

plt.figure(figsize=(7,4.5))
for B in B_list:
    N, Nu, Nd, meta = oddfreq_dos_with_fm(E, B, InOx_params, fm_window=fm_window)
    plt.plot(E, Nu, label=f"↑, B={B}T")
    plt.plot(E, Nd, label=f"↓, B={B}T")
plt.title("Spin-resolved DOS with a hypothesized emergent FM window (toy)")
plt.xlabel("Energy (eV)"); plt.ylabel("N(E)/N0")
plt.legend(); plt.tight_layout(); plt.show()

# Export a representative field
B_rep = 9.5
N, Nu, Nd, meta = oddfreq_dos_with_fm(E, B_rep, InOx_params, fm_window=fm_window)
pd.DataFrame({"E_eV": E, "N_total": N, "N_up": Nu, "N_dn": Nd}).to_csv("/content/dos/InOx_odd_with_FMwindow_B9p5T.csv", index=False)
meta
```

```python
# Parameter fitter for odd-ω InOx model + presets for NbN and MoGe, and cross-material plots.
# This cell adds:
#  1) A simple least-squares fitter (random-restart grid) to match N(0)(B) or R/Rn(B).
#  2) Material presets for NbN and MoGe.
#  3) Demo runs for both materials (DOS overlays) and an example fit using a synthetic target.
#
# Notes:
#  - Replace `target_csv_path` with your data file path (columns: B_T plus either N0 or R_over_Rn).
#  - Choose which parameters to fit and provide sensible bounds.
#  - The fitter is gradient-free and robust to modest noise.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ---- Reuse odd-frequency machinery from previous cell ----
try:
    InOx_params
    oddfreq_dos
    plot_odd_freq_overlays
except NameError:
    # Minimal redefinitions if needed
    mu_B_eV_T = 5.7883818060e-5
    g_eff = 2.0
    def Delta_even(B, p):
        x = max(0.0, 1.0 - (B/p["Bc2_T"])**2); return p["Delta0_eV"] * np.sqrt(x)
    def eta_odd(B, p):
        base = p["eta0"] * (B/p["Bc2_T"])**p["eta_power"]
        if B <= p["Bc2_T"]:
            return base
        x = min(1.0, max(0.0, 1.0 - (B/p["Bc2_T"] - 1.0)/(p.get("eta_tail_stop_frac",1.1) - 1.0)))
        return base * (x**p.get("eta_tail_shape",2.0))
    def Gamma_tot(B, p):
        Δ0 = p["Delta0_eV"]
        return p["gamma_dynes_frac"]*Δ0 + p["Gamma_pb0_frac"]*Δ0*(B/p["Bc2_T"])**2
    def zeeman_h(B, p):
        if not p.get("include_zeeman", True): return 0.0
        return p.get("zeeman_scale", 0.25) * (g_eff * mu_B_eV_T * B / 2.0)
    def oddfreq_dos(E, B, p):
        Δe = Delta_even(B, p); η = eta_odd(B, p); Γ = Gamma_tot(B, p); h = zeeman_h(B, p)
        z_up = (E + h) + 1j*Γ; z_dn = (E - h) + 1j*Γ
        Δeff_up = Δe + 1j*η*(E + h); Δeff_dn = Δe + 1j*η*(E - h)
        Nu = np.abs(np.real(z_up / np.sqrt(z_up**2 - Δeff_up**2)))
        Nd = np.abs(np.real(z_dn / np.sqrt(z_dn**2 - Δeff_dn**2)))
        return 0.5*(Nu+Nd), Nu, Nd, dict(Delta_even=Δe, eta=η, Gamma=Γ, h=h)
    def plot_odd_freq_overlays(B_list, p, Emin=-0.004, Emax=0.004, nE=2500, prefix="oddmodel"):
        E = np.linspace(Emin, Emax, nE)
        plt.figure(figsize=(7,4.5))
        for B in B_list:
            N,_,_,_ = oddfreq_dos(E, B, p)
            plt.plot(E, N, label=f"B={B:.1f} T")
            pd.DataFrame({"E_eV":E, "N_over_N0":N}).to_csv(f"/content/dos/{prefix}_DOS_B{B:.1f}T.csv", index=False)
        plt.title("DOS with odd-ω mixing vs field (toy)"); plt.xlabel("Energy (eV)"); plt.ylabel("N(E)/N0"); plt.legend(); plt.tight_layout(); plt.show()
        # N(0) vs B
        N0s = []
        for B in B_list:
            N,_,_,_ = oddfreq_dos(np.array([0.0]), B, p)
            N0s.append(float(N[0]))
        df = pd.DataFrame({"B_T":B_list,"N0":N0s})
        df.to_csv(f"/content/dos/{prefix}_summary.csv", index=False)
        plt.figure(figsize=(7,4.5)); plt.plot(B_list, N0s, marker="o")
        plt.title("N(0)/N0 vs B"); plt.xlabel("B (T)"); plt.ylabel("N(0)/N0"); plt.tight_layout(); plt.show()

# ---- Utilities ----
def N0_vs_B(B_list, p):
    """Compute zero-energy DOS for a parameter dict p."""
    N0 = []
    for B in B_list:
        N,_,_,_ = oddfreq_dos(np.array([0.0]), B, p)
        N0.append(float(N[0]))
    return np.array(N0)

def R_over_Rn_from_N0(N0, eps_floor=0.02):
    """Toy mapping from N0 to reduced resistance; normalize to 1 at highest B."""
    R = 1.0 / (eps_floor + N0)
    return R / R[-1]

def fit_odd_params(target_csv_path,
                   y_col="N0",
                   B_col="B_T",
                   base_params=None,
                   vary=("eta0","eta_power","Gamma_pb0_frac","zeeman_scale","Bc2_T","gamma_dynes_frac"),
                   bounds=None,
                   eps_floor=0.02,
                   rounds=(800, 400, 200),
                   shrink=0.35,
                   seed=0,
                   export_prefix="fit"):
    """
    Fit odd-ω model parameters to target data y(B).
    - y_col: 'N0' or 'R_over_Rn'
    - bounds: dict {param: (lo, hi)}; any not provided get +/-50% around base value (except exponents, clamp >=0)
    - rounds: random samples per stage; we shrink bounds around current best each stage.
    Returns (best_params, fit_df, summary_dict).
    """
    rng = np.random.default_rng(seed)
    # Load target
    tgt = pd.read_csv(target_csv_path)
    B = tgt[B_col].values.astype(float)
    y = tgt[y_col].values.astype(float)
    # Base params
    if base_params is None:
        p0 = dict(InOx_params)  # copy InOx baseline
    else:
        p0 = dict(base_params)
    # Build bounds
    bnds = {}
    for k in vary:
        if bounds and k in bounds:
            bnds[k] = list(bounds[k])
        else:
            v = p0[k]
            if k in ("eta_power",):
                lo, hi = max(0.0, 0.5*v), max(0.6, 1.5*v)
            elif k in ("Bc2_T",):
                lo, hi = 0.5*v, 2.0*v
            else:
                lo, hi = max(1e-6, 0.5*v), 1.5*v
            bnds[k] = [lo, hi]

    def evaluate(pdict):
        N0 = N0_vs_B(B, pdict)
        if y_col.lower().startswith("r"):
            yhat = R_over_Rn_from_N0(N0, eps_floor=eps_floor)
            # scale factor to match arbitrary normalization (if any)
            s = (y @ yhat) / (yhat @ yhat) if (yhat @ yhat) > 0 else 1.0
            yhat = s * yhat
        else:
            yhat = N0
        err = yhat - y
        return float(np.mean(err**2)), yhat

    best_p = dict(p0)
    best_loss, best_yhat = evaluate(best_p)

    # Random-restart shrinking box search
    for stage, nsamp in enumerate(rounds):
        for _ in range(nsamp):
            cand = dict(best_p)
            for k in vary:
                lo, hi = bnds[k]
                cand[k] = float(rng.uniform(lo, hi))
            loss, _ = evaluate(cand)
            if loss < best_loss:
                best_loss, best_p = loss, cand
        # shrink bounds around best
        for k in vary:
            lo, hi = bnds[k]
            mid = best_p[k]
            width = (hi - lo) * shrink
            bnds[k] = [max(1e-12, mid - width/2), mid + width/2]

    # Final evaluation
    loss, yhat = evaluate(best_p)
    out_df = pd.DataFrame({B_col: B, "y_target": y, "y_model": yhat})
    out_df.to_csv(f"/content/dos/{export_prefix}_fit_curve.csv", index=False)
    pd.DataFrame([best_p]).to_json(f"/content/dos/{export_prefix}_best_params.json", orient="records", indent=2)

    # Plot
    plt.figure(figsize=(7,4.5))
    plt.plot(B, y, marker="o", linestyle="none", label="Target")
    plt.plot(B, yhat, label="Model fit")
    plt.title(f"Odd-ω model fit to {y_col}(B)  (MSE={loss:.3g})")
    plt.xlabel("B (T)"); plt.ylabel(y_col); plt.legend(); plt.tight_layout(); plt.show()

    return best_p, out_df, {"mse": loss, "export_curve": f"/content/dos/{export_prefix}_fit_curve.csv",
                            "export_params": f"/content/dos/{export_prefix}_best_params.json"}

# ---- Presets for additional materials ----
kB = 8.617333262145e-5
NbN_params = dict(
    Tc_K = 15.0,
    Delta0_eV = 1.76*kB*15.0,   # ~2.27 meV
    Bc2_T = 25.0,
    gamma_dynes_frac = 0.02,
    Gamma_pb0_frac = 0.02,
    eta0 = 0.4,
    eta_power = 1.0,
    include_zeeman = True,
    zeeman_scale = 0.2,
    eta_tail_stop_frac = 1.1,
    eta_tail_shape = 2.0
)

MoGe_params = dict(
    Tc_K = 8.0,
    Delta0_eV = 1.76*kB*8.0,   # ~1.21 meV
    Bc2_T = 12.0,
    gamma_dynes_frac = 0.025,
    Gamma_pb0_frac = 0.03,
    eta0 = 0.6,
    eta_power = 1.1,
    include_zeeman = True,
    zeeman_scale = 0.25,
    eta_tail_stop_frac = 1.1,
    eta_tail_shape = 2.0
)

# ---- Cross-material comparisons (demo) ----
B_demo = [0.0, 0.3*NbN_params["Bc2_T"], 0.6*NbN_params["Bc2_T"], 0.9*NbN_params["Bc2_T"], 1.1*NbN_params["Bc2_T"]]
plot_odd_freq_overlays(B_demo, p=NbN_params, Emin=-0.006, Emax=0.006, nE=2500, prefix="NbN_odd")
B_demo2 = [0.0, 0.3*MoGe_params["Bc2_T"], 0.6*MoGe_params["Bc2_T"], 0.9*MoGe_params["Bc2_T"], 1.1*MoGe_params["Bc2_T"]]
plot_odd_freq_overlays(B_demo2, p=MoGe_params, Emin=-0.005, Emax=0.005, nE=2500, prefix="MoGe_odd")

# ---- Example fit with synthetic target (so you can see it working) ----
# Generate a synthetic "experiment" from InOx baseline then add noise; fit back.
B_syn = np.array([0.0, 2.0, 4.0, 6.0, 7.5, 8.5, 9.0, 9.5, 10.0, 11.0])
N0_syn = N0_vs_B(B_syn, InOx_params)
# make a toy R/Rn signal (pretend experiment reports R/Rn)
R_syn = R_over_Rn_from_N0(N0_syn, eps_floor=0.02)
rng = np.random.default_rng(42)
R_syn_noisy = np.clip(R_syn + rng.normal(0, 0.03, size=R_syn.size), 0, None)  # 3% noise

syn_path = "/content/dos/synthetic_InOx_target.csv"
pd.DataFrame({"B_T": B_syn, "R_over_Rn": R_syn_noisy}).to_csv(syn_path, index=False)

# Fit a subset of parameters commonly tied to field-odd mixing
vary_keys = ("eta0","eta_power","Gamma_pb0_frac","Bc2_T","gamma_dynes_frac","zeeman_scale")
bounds = {
    "eta0": (0.05, 1.5),
    "eta_power": (0.5, 2.0),
    "Gamma_pb0_frac": (0.0, 0.08),
    "Bc2_T": (6.0, 20.0),
    "gamma_dynes_frac": (0.005, 0.06),
    "zeeman_scale": (0.0, 0.5)
}

best_params, fit_df, fit_info = fit_odd_params(syn_path,
                                               y_col="R_over_Rn",
                                               base_params=InOx_params,
                                               vary=vary_keys,
                                               bounds=bounds,
                                               eps_floor=0.02,
                                               rounds=(600, 300, 200),
                                               shrink=0.4,
                                               seed=1,
                                               export_prefix="InOx_synth_fit")

(best_params, fit_info)
```

```python
# Patch the Majorana simulator: fix antisymmetrizer dimension for Pfaffian (use U = i τ_y ⊗ I_spin).
# Recompute phase diagram, spectrum, and LDOS.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Recreate matrices (in case of fresh kernel)
sx = np.array([[0,1],[1,0]], dtype=complex)
sy = np.array([[0,-1j],[1j,0]], dtype=complex)
sz = np.array([[1,0],[0,-1]], dtype=complex)
s0 = np.eye(2, dtype=complex)

tx = np.array([[0,1],[1,0]], dtype=complex)
ty = np.array([[0,-1j],[1j,0]], dtype=complex)
tz = np.array([[1,0],[0,-1]], dtype=complex)
t0 = np.eye(2, dtype=complex)

def kron(a,b):
    return np.kron(a,b)

def hk_1d(k, t, mu, alpha, Delta, Vz):
    xi = -2*t*np.cos(k) - mu
    H = xi * kron(tz, s0) + alpha*np.sin(k)*kron(tz, sy) + Vz*kron(t0, sx) + Delta*kron(tx, s0)
    return H

def pfaffian(A):
    A = np.array(A, dtype=complex)
    n = A.shape[0]
    if n == 0:
        return 1.0 + 0j
    if n % 2 == 1:
        return 0.0 + 0j
    for j in range(1, n):
        if abs(A[0, j]) > 1e-18:
            break
    else:
        return 0.0 + 0j
    sub = np.delete(np.delete(A, [0, j], axis=0), [0, j], axis=1)
    sign = (-1)**(j+1)
    return sign * A[0, j] * pfaffian(sub)

def majorana_number_Z2(t, mu, alpha, Delta, Vz):
    # Use U = i τ_y ⊗ I_spin so that B(k) = H(k) U is antisymmetric at k=0,π
    U = 1j * kron(ty, s0)
    H0 = hk_1d(0.0, t, mu, alpha, Delta, Vz)
    Hpi = hk_1d(np.pi, t, mu, alpha, Delta, Vz)
    B0 = 0.5*(H0 @ U - (H0 @ U).T)
    Bpi = 0.5*(Hpi @ U - (Hpi @ U).T)
    pf0 = pfaffian(B0)
    pfpi = pfaffian(Bpi)
    M = np.sign(np.real(pf0*pfpi))
    return int(M), pf0, pfpi

def bulk_gap(t, mu, alpha, Delta, Vz, nk=401):
    ks = np.linspace(0, np.pi, nk)
    gap = np.inf
    for k in ks:
        w = np.linalg.eigvalsh(hk_1d(k, t, mu, alpha, Delta, Vz))
        pos = np.sort(np.abs(w))
        gap = min(gap, pos[0])
    return float(gap)

def phase_diagram(mu_vals, Vz_vals, t, alpha, Delta, nk_gap=201):
    Mgrid = np.zeros((len(Vz_vals), len(mu_vals)))
    Gap = np.zeros_like(Mgrid)
    for i, Vz in enumerate(Vz_vals):
        for j, mu in enumerate(mu_vals):
            M, _, _ = majorana_number_Z2(t, mu, alpha, Delta, Vz)
            Mgrid[i, j] = M
            Gap[i, j] = bulk_gap(t, mu, alpha, Delta, Vz, nk=nk_gap)
    df = pd.DataFrame({"Vz_eV": np.repeat(Vz_vals, len(mu_vals)),
                       "mu_eV": np.tile(mu_vals, len(Vz_vals)),
                       "Z2": Mgrid.flatten(),
                       "gap_eV": Gap.flatten()})
    path = "/content/2majorana_phase_diagram.csv"
    df.to_csv(path, index=False)

    plt.figure(figsize=(7,4.5))
    plt.imshow(Mgrid, origin="lower",
               extent=[mu_vals[0], mu_vals[-1], Vz_vals[0], Vz_vals[-1]],
               aspect="auto", vmin=-1, vmax=1)
    plt.title("Z2 Majorana number (class D): -1 = topological")
    plt.xlabel("μ (eV)"); plt.ylabel("V_z (eV)")
    plt.tight_layout(); plt.show()

    plt.figure(figsize=(7,4.5))
    plt.imshow(Gap, origin="lower",
               extent=[mu_vals[0], mu_vals[-1], Vz_vals[0], Vz_vals[-1]],
               aspect="auto")
    plt.title("Bulk gap (eV)")
    plt.xlabel("μ (eV)"); plt.ylabel("V_z (eV)")
    plt.tight_layout(); plt.show()
    return path

def build_wire_H(N, t, mu, alpha, Delta, Vz):
    dim = 4*N
    H = np.zeros((dim, dim), dtype=complex)
    h0 = (2*t - mu)*kron(tz, np.eye(2)) + Vz*kron(t0, sx) + Delta*kron(tx, np.eye(2))
    hop = (-t)*kron(tz, np.eye(2)) + (1j*alpha/2.0)*kron(tz, sy)
    for i in range(N):
        si = slice(4*i, 4*(i+1))
        H[si, si] = h0
        if i < N-1:
            sj = slice(4*(i+1), 4*(i+2))
            H[si, sj] = hop
            H[sj, si] = hop.conj().T
    return H

def spectrum_vs_Vz(N, t, mu, alpha, Delta, Vz_vals, n_show=8):
    energies = []
    for Vz in Vz_vals:
        H = build_wire_H(N, t, mu, alpha, Delta, Vz)
        w = np.linalg.eigvalsh(H)
        w = np.sort(w, key=lambda x: abs(x))
        energies.append(w[:n_show])
    energies = np.array(energies)
    df = pd.DataFrame({"Vz_eV": Vz_vals})
    for n in range(n_show):
        df[f"E{n}_eV"] = energies[:, n]
    path = "/content/2majorana_wire_spectrum.csv"
    df.to_csv(path, index=False)

    plt.figure(figsize=(7,4.5))
    for n in range(n_show):
        plt.plot(Vz_vals, energies[:, n])
    plt.axhline(0.0, linestyle="--")
    plt.title("Finite wire spectrum vs Vz (look for zero modes)")
    plt.xlabel("V_z (eV)"); plt.ylabel("Energy (eV)")
    plt.tight_layout(); plt.show()
    return path

def end_mode_ldos(N, t, mu, alpha, Delta, Vz, n_modes=2, e_cut=1e-6):
    H = build_wire_H(N, t, mu, alpha, Delta, Vz)
    w, V = np.linalg.eigh(H)
    idx = np.argsort(np.abs(w))
    w = w[idx]; V = V[:, idx]
    chosen = []
    for n in range(len(w)):
        if len(chosen) >= n_modes: break
        if abs(w[n]) <= e_cut:
            chosen.append(n)
    if not chosen:
        chosen = list(range(n_modes))
    weights = np.zeros((len(chosen), N))
    for m, n in enumerate(chosen):
        psi = V[:, n]
        for i in range(N):
            si = slice(4*i, 4*(i+1))
            weights[m, i] = float(np.sum(np.abs(psi[si])**2))

    plt.figure(figsize=(7,4.5))
    for m in range(weights.shape[0]):
        plt.plot(np.arange(N), weights[m], label=f"mode {m}  E≈{w[chosen[m]]:.1e} eV")
    plt.title("Near-zero-mode spatial weight (end-localized Majoranas)")
    plt.xlabel("site i"); plt.ylabel("|ψ|² (arb.)")
    plt.legend(); plt.tight_layout(); plt.show()

    df = pd.DataFrame(weights.T, columns=[f"mode{m}" for m in range(weights.shape[0])])
    df.insert(0, "site", np.arange(N))
    path = "/content/2majorana_endmode_weights.csv"
    df.to_csv(path, index=False)
    return path

# Presets
presets = {
    "EuS|Al_wire": dict(t=0.010, alpha=0.003, Delta=0.00018, mu=0.0),
    "InOx|FM_wire": dict(t=0.010, alpha=0.002, Delta=1.76*8.617333262145e-5*3.0, mu=0.0),
    "UGe2_like_wire": dict(t=0.012, alpha=0.0025, Delta=1.76*8.617333262145e-5*0.8, mu=0.0)
}

# Run on EuS|Al preset
cfg = presets["EuS|Al_wire"]
t = cfg["t"]; alpha = cfg["alpha"]; Delta = cfg["Delta"]; mu0 = cfg["mu"]

mu_vals = np.linspace(-0.004, 0.004, 121)
Vz_vals = np.linspace(0.0, 0.004, 121)
phase_csv = phase_diagram(mu_vals, Vz_vals, t, alpha, Delta, nk_gap=121)

N = 100
Vz_sweep = np.linspace(0.0, 0.004, 121)
spec_csv = spectrum_vs_Vz(N, t, mu0, alpha, Delta, Vz_sweep, n_show=12)

Vz_topo = 0.0015
ldos_csv = end_mode_ldos(N, t, mu0, alpha, Delta, Vz_topo, n_modes=2, e_cut=5e-6)

[phase_csv, spec_csv, ldos_csv]
```

```python
# Smooth confinement + odd-ω InOx self-energy for the Majorana nanowire.
# - Smooth confinement: soft edge potential V_conf(i) added to μ_i.
# - Odd-ω self-energy (InOx): Σ(E) = -i Γ(B) * I  +  i η(B) E * τ_x (applied to a left-interface region with decay).
# - Compute LDOS at the left end and mid-wire using Green's functions with energy-dependent Σ.
#
# This is a toy-yet-useful phenomenology to see how soft boundaries and odd-ω proximity
# shape end states and zero-bias spectral weight.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ---------- Pauli blocks ----------
sx = np.array([[0,1],[1,0]], dtype=complex)
sy = np.array([[0,-1j],[1j,0]], dtype=complex)
sz = np.array([[1,0],[0,-1]], dtype=complex)
s0 = np.eye(2, dtype=complex)

tx = np.array([[0,1],[1,0]], dtype=complex)
ty = np.array([[0,-1j],[1j,0]], dtype=complex)
tz = np.array([[1,0],[0,-1]], dtype=complex)
t0 = np.eye(2, dtype=complex)

def kron(a,b): return np.kron(a,b)

# ---------- Smooth confinement ----------
def smooth_confinement(N, V0=0.0006, w=8, span=3):
    """
    Soft edge potential that raises μ near the ends.
    V_conf[i] ~ V0 near i~0 and i~N-1; ~0 in the bulk.  'w' sets edge softness, 'span' multiplies w.
    """
    i = np.arange(N)
    left = 1.0/(1.0 + np.exp((i - span*w)/w))
    right = 1.0/(1.0 + np.exp(((N-1 - i) - span*w)/w))
    return V0 * (left + right)

# ---------- Wire Hamiltonian with site dependence ----------
def build_wire_H_smooth(N, t, mu0, alpha, Delta, Vz, Vconf=None, Delta_profile=None):
    """
    Real-space 4N x 4N BdG with site-dependent μ_i and Δ_i and smooth confinement.
    """
    if Vconf is None:
        Vconf = np.zeros(N)
    if Delta_profile is None:
        Delta_profile = np.full(N, Delta)

    dim = 4*N
    H = np.zeros((dim, dim), dtype=complex)

    # On-site block at site i: (2t - μ_i) τ_z + Vz σ_x + Δ_i τ_x
    for i in range(N):
        mu_i = mu0 + Vconf[i]
        h_onsite = (2*t - mu_i)*kron(tz, s0) + Vz*kron(t0, sx) + Delta_profile[i]*kron(tx, s0)
        si = slice(4*i, 4*(i+1))
        H[si, si] = h_onsite

    # Hopping: (-t) τ_z + (i α /2) τ_z σ_y
    hop = (-t)*kron(tz, s0) + (1j*alpha/2.0)*kron(tz, sy)
    for i in range(N-1):
        si = slice(4*i, 4*(i+1))
        sj = slice(4*(i+1), 4*(i+2))
        H[si, sj] = hop
        H[sj, si] = hop.conj().T

    return H

# ---------- InOx odd-ω self-energy ----------
# Defaults (can be overridden)
InOx_params = dict(
    Tc_K = 3.0,
    Delta0_eV = 1.76 * 8.617333262145e-5 * 3.0,  # ≈0.000455 eV
    Bc2_T = 10.0,
    gamma_dynes_frac = 0.02,
    Gamma_pb0_frac = 0.03,
    eta0 = 0.8,
    eta_power = 1.2,
    eta_tail_stop_frac = 1.1,
    eta_tail_shape = 2.0
)

mu_B_eV_T = 5.7883818060e-5
g_eff = 2.0

def Delta_even(B, p): return p["Delta0_eV"] * np.sqrt(max(0.0, 1.0 - (B/p["Bc2_T"])**2))
def Gamma_tot(B, p):
    Δ0 = p["Delta0_eV"]
    return p["gamma_dynes_frac"]*Δ0 + p["Gamma_pb0_frac"]*Δ0*(B/p["Bc2_T"])**2
def eta_odd(B, p):
    base = p["eta0"] * (B/p["Bc2_T"])**p["eta_power"]
    if B <= p["Bc2_T"]: return base
    x = min(1.0, max(0.0, 1.0 - (B/p["Bc2_T"] - 1.0)/(p["eta_tail_stop_frac"] - 1.0)))
    return base * (x**p["eta_tail_shape"])

def make_sigma_inox(N, E, B, p, left_len=12, xi=6.0, scale=1.0):
    """
    Build Σ(E) for the wire due to an InOx contact covering the first 'left_len' sites, decaying as exp(-i/xi).
    Σ(E) = -i Γ(B) I_4 + i η(B) E τ_x (⊗ I_spin), per site, weighted by w_i = scale * exp(-i/xi).
    """
    Γ = Gamma_tot(B, p)
    η = eta_odd(B, p)
    dim = 4*N
    Sig = np.zeros((dim, dim), dtype=complex)
    # local building blocks
    diag_block = -1j*Γ * np.eye(4, dtype=complex)
    off_block  = 1j*η*E * kron(tx, s0)
    for i in range(min(N, left_len)):
        w = scale * np.exp(-i/xi)
        si = slice(4*i, 4*(i+1))
        Sig[si, si] += w * (diag_block + off_block)
    return Sig

# ---------- LDOS with Σ(E) ----------
def ldos_with_selfenergy(N, H0, E_list, Sigma_fn, sites=("end","mid")):
    """
    Compute electron LDOS: ρ_i(E) = -(1/π) Im Tr_spin G_ee^R(i,i;E).
    Sigma_fn(E) returns a 4N x 4N matrix Σ(E).
    """
    dim = 4*N
    # electron projector for site i
    def proj_indices(i):
        base = 4*i
        return [base+0, base+1]  # electron ↑,↓ (first two of 4)
    # site indices to probe
    idxs = {}
    if "end" in sites: idxs["end"] = proj_indices(0)
    if "mid" in sites: idxs["mid"] = proj_indices(N//2)
    if "far" in sites: idxs["far"] = proj_indices(N-1)

    out = {k: np.zeros(len(E_list)) for k in idxs.keys()}
    eye = np.eye(dim, dtype=complex)
    eta_small = 1e-7  # infinitesimal
    for m, E in enumerate(E_list):
        A = (E + 1j*eta_small)*eye - H0 - Sigma_fn(E)
        # Full inverse (smallish system) — robust & simple
        G = np.linalg.inv(A)
        for tag, inds in idxs.items():
            # sum Im of diagonal elements of electron block
            val = 0.0
            for j in inds:
                val += -np.imag(G[j, j]) / np.pi
            out[tag][m] = float(val)
    return out

# ---------- Demo: EuS|InAs with smooth confinement + InOx Σ ----------
# Wire parameters (EuS|InAs-like)
t, alpha, Delta, mu0 = 0.010, 0.0045, 0.00020, 0.0
N = 60
Vz = 1.5*Delta  # pick a point near/inside topo window
# Smooth confinement
Vconf = smooth_confinement(N, V0=4*Delta, w=8, span=3)
# Build H0
H0 = build_wire_H_smooth(N, t, mu0, alpha, Delta, Vz, Vconf=Vconf, Delta_profile=None)

# Self-energy config (InOx contact on the left)
B_field = 9.0  # Tesla, for odd-ω strength; tweakable
left_len = 14  # number of sites under the InOx "cap"
xi_inox = 6.0  # decay length in sites
scale_coupling = 1.0

def Sigma_E(E):
    return make_sigma_inox(N, E, B_field, InOx_params, left_len=left_len, xi=xi_inox, scale=scale_coupling)

# Energy grid and LDOS
E_list = np.linspace(-4*Delta, 4*Delta, 181)
ldos = ldos_with_selfenergy(N, H0, E_list, Sigma_E, sites=("end","mid","far"))

# Plots
plt.figure(figsize=(7,4.5))
plt.plot(E_list, ldos["end"], label="end")
plt.plot(E_list, ldos["mid"], label="mid")
plt.plot(E_list, ldos["far"], label="far end")
plt.title("LDOS with smooth confinement + InOx odd-ω self-energy (EuS|InAs)")
plt.xlabel("Energy (eV)"); plt.ylabel("LDOS (arb.)")
plt.legend(); plt.tight_layout(); plt.show()

# Export CSVs
df_end = pd.DataFrame({"E_eV": E_list, "LDOS_end": ldos["end"]})
df_mid = pd.DataFrame({"E_eV": E_list, "LDOS_mid": ldos["mid"]})
df_far = pd.DataFrame({"E_eV": E_list, "LDOS_far": ldos["far"]})
p_end = "/content/LDOS_end_smooth_inox.csv"
p_mid = "/content/LDOS_mid_smooth_inox.csv"
p_far = "/content/LDOS_far_smooth_inox.csv"
df_end.to_csv(p_end, index=False)
df_mid.to_csv(p_mid, index=False)
df_far.to_csv(p_far, index=False)

[p_end, p_mid, p_far]
```

```python
# Sweep magnetic field B in the InOx self-energy and compute LDOS (end/mid/far).
# Uses the optimized solver path (linear solves on selected indices).
# Produces: (1) heatmaps (E vs B) for end/mid/far, (2) N(0) vs B, and (3) a tidy CSV export.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ---------- Pauli blocks & helpers (redeclare to be safe) ----------
sx = np.array([[0,1],[1,0]], dtype=complex)
sy = np.array([[0,-1j],[1j,0]], dtype=complex)
s0 = np.eye(2, dtype=complex)
tx = np.array([[0,1],[1,0]], dtype=complex)
ty = np.array([[0,-1j],[1j,0]], dtype=complex)
tz = np.array([[1,0],[0,-1]], dtype=complex)
t0 = np.eye(2, dtype=complex)
def kron(a,b): return np.kron(a,b)

def smooth_confinement(N, V0=0.0006, w=6, span=3):
    i = np.arange(N)
    left = 1.0/(1.0 + np.exp((i - span*w)/w))
    right = 1.0/(1.0 + np.exp(((N-1 - i) - span*w)/w))
    return V0 * (left + right)

def build_wire_H_smooth(N, t, mu0, alpha, Delta, Vz, Vconf=None, Delta_profile=None):
    if Vconf is None: Vconf = np.zeros(N)
    if Delta_profile is None: Delta_profile = np.full(N, Delta)
    dim = 4*N
    H = np.zeros((dim, dim), dtype=complex)
    for i in range(N):
        mu_i = mu0 + Vconf[i]
        h_onsite = (2*t - mu_i)*kron(tz, s0) + Vz*kron(t0, sx) + Delta_profile[i]*kron(tx, s0)
        si = slice(4*i, 4*(i+1)); H[si, si] = h_onsite
    hop = (-t)*kron(tz, s0) + (1j*alpha/2.0)*kron(tz, sy)
    for i in range(N-1):
        si = slice(4*i, 4*(i+1)); sj = slice(4*(i+1), 4*(i+2))
        H[si, sj] = hop; H[sj, si] = hop.conj().T
    return H

# InOx odd-ω self-energy params/functions
InOx_params = dict(
    Tc_K = 3.0,
    Delta0_eV = 1.76 * 8.617333262145e-5 * 3.0,
    Bc2_T = 10.0,
    gamma_dynes_frac = 0.02,
    Gamma_pb0_frac = 0.03,
    eta0 = 0.8,
    eta_power = 1.2,
    eta_tail_stop_frac = 1.1,
    eta_tail_shape = 2.0
)
def Gamma_tot(B, p):
    Δ0 = p["Delta0_eV"]
    return p["gamma_dynes_frac"]*Δ0 + p["Gamma_pb0_frac"]*Δ0*(B/p["Bc2_T"])**2
def eta_odd(B, p):
    base = p["eta0"] * (B/p["Bc2_T"])**p["eta_power"]
    if B <= p["Bc2_T"]: return base
    x = min(1.0, max(0.0, 1.0 - (B/p["Bc2_T"] - 1.0)/(p["eta_tail_stop_frac"] - 1.0)))
    return base * (x**p["eta_tail_shape"])

def make_sigma_inox(N, E, B, p, left_len=12, xi=6.0, scale=1.0):
    Γ = Gamma_tot(B, p); η = eta_odd(B, p)
    dim = 4*N; Sig = np.zeros((dim, dim), dtype=complex)
    diag_block = -1j*Γ * np.eye(4, dtype=complex)
    off_block  = 1j*η*E * kron(tx, s0)
    for i in range(min(N, left_len)):
        w = scale * np.exp(-i/xi)
        si = slice(4*i, 4*(i+1))
        Sig[si, si] += w * (diag_block + off_block)
    return Sig

def ldos_sites_with_selfenergy(N, H0, E_list, Sigma_fn, site_indices):
    dim = 4*N; eye = np.eye(dim, dtype=complex); eta_small = 1e-7
    ldos = {name: np.zeros(len(E_list)) for name in site_indices.keys()}
    for m, E in enumerate(E_list):
        A = (E + 1j*eta_small)*eye - H0 - Sigma_fn(E)
        # compute selected diagonals via linear solves
        for name, idxs in site_indices.items():
            val = 0.0
            for j in idxs:
                e = np.zeros(dim, dtype=complex); e[j] = 1.0
                x = np.linalg.solve(A, e)  # column j of G
                val += -np.imag(x[j]) / np.pi
            ldos[name][m] = float(val)
    return ldos

# ---------- Setup wire & sweep ----------
# Wire (EuS|InAs-like), soft edges
t, alpha, Delta, mu0 = 0.010, 0.0045, 0.00020, 0.0
N = 40
Vz = 1.5*Delta
Vconf = smooth_confinement(N, V0=4*Delta, w=6, span=3)
H0 = build_wire_H_smooth(N, t, mu0, alpha, Delta, Vz, Vconf=Vconf, Delta_profile=None)

# InOx contact geometry
left_len = 12; xi_inox = 5.0; scale_coupling = 1.0

# Energies & B sweep
E_list = np.linspace(-4*Delta, 4*Delta, 101)
B_list = np.array([0.0, 3.0, 6.0, 8.0, 9.0, 9.5, 10.0, 11.0, 12.0])  # crosses Bc2=10 T

# Sites to record
def e_inds(site): base = 4*site; return [base+0, base+1]
site_indices = {"end": e_inds(0), "mid": e_inds(N//2), "far": e_inds(N-1)}

# Accumulators
LDOS_end = np.zeros((len(E_list), len(B_list)))
LDOS_mid = np.zeros_like(LDOS_end)
LDOS_far = np.zeros_like(LDOS_end)
N0_end = np.zeros(len(B_list)); N0_mid = np.zeros(len(B_list)); N0_far = np.zeros(len(B_list))

# Sweep B
for k, B in enumerate(B_list):
    def Sigma_E(E, Bval=B):
        return make_sigma_inox(N, E, Bval, InOx_params, left_len=left_len, xi=xi_inox, scale=scale_coupling)
    ldos = ldos_sites_with_selfenergy(N, H0, E_list, Sigma_E, site_indices)
    LDOS_end[:, k] = ldos["end"]
    LDOS_mid[:, k] = ldos["mid"]
    LDOS_far[:, k] = ldos["far"]
    # Zero-bias (E=0) LDOS
    N0_end[k] = float(np.interp(0.0, E_list, ldos["end"]))
    N0_mid[k] = float(np.interp(0.0, E_list, ldos["mid"]))
    N0_far[k] = float(np.interp(0.0, E_list, ldos["far"]))

# ---------- Plots: heatmaps & N(0) vs B ----------
extent = [B_list[0], B_list[-1], E_list[0], E_list[-1]]

plt.figure(figsize=(7,4.5))
plt.imshow(LDOS_end, origin="lower", extent=extent, aspect="auto")
plt.title("End LDOS vs B and E (EuS|InAs + InOx Σ)")
plt.xlabel("B (T)"); plt.ylabel("Energy (eV)")
plt.tight_layout(); plt.show()

plt.figure(figsize=(7,4.5))
plt.imshow(LDOS_mid, origin="lower", extent=extent, aspect="auto")
plt.title("Mid LDOS vs B and E")
plt.xlabel("B (T)"); plt.ylabel("Energy (eV)")
plt.tight_layout(); plt.show()

plt.figure(figsize=(7,4.5))
plt.imshow(LDOS_far, origin="lower", extent=extent, aspect="auto")
plt.title("Far-end LDOS vs B and E")
plt.xlabel("B (T)"); plt.ylabel("Energy (eV)")
plt.tight_layout(); plt.show()

plt.figure(figsize=(7,4.5))
plt.plot(B_list, N0_end, marker="o", label="end")
plt.plot(B_list, N0_mid, marker="s", label="mid")
plt.plot(B_list, N0_far, marker="^", label="far")
plt.axvline(10.0, linestyle="--")  # Bc2
plt.title("Zero-bias LDOS N(0) vs B")
plt.xlabel("B (T)"); plt.ylabel("LDOS at E=0 (arb.)")
plt.legend(); plt.tight_layout(); plt.show()

# ---------- Export tidy CSV ----------
rows = []
for j, B in enumerate(B_list):
    for i, E in enumerate(E_list):
        rows.append({"B_T": B, "E_eV": E, "LDOS_end": LDOS_end[i,j], "LDOS_mid": LDOS_mid[i,j], "LDOS_far": LDOS_far[i,j]})
tidy = pd.DataFrame(rows)
path_csv = "/content/LDOS_vsB_EuS_InAs_InOxSigma_tidy.csv"
tidy.to_csv(path_csv, index=False)

# Also export N(0) vs B
pd.DataFrame({"B_T": B_list, "N0_end": N0_end, "N0_mid": N0_mid, "N0_far": N0_far}).to_csv("/content/N0_vsB_EuS_InAs_InOxSigma.csv", index=False)

[path_csv, "/content/N0_vsB_EuS_InAs_InOxSigma.csv"]
```
