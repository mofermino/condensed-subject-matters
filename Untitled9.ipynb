```python
# Create a side-by-side ISI vs Hofstadter-style diagram using matplotlib (no seaborn, default colors)
import numpy as np
import matplotlib.pyplot as plt

# --- Helper: Gaussian pulse ---
def gaussian(t, mu, sigma):
    return np.exp(-0.5*((t-mu)/sigma)**2)

# Build figure
fig = plt.figure(figsize=(12, 5), dpi=150)
gs = fig.add_gridspec(1, 2, width_ratios=[1,1], wspace=0.25)

# ===== Left panel: ISI schematic =====
ax1 = fig.add_subplot(gs[0,0])

t = np.linspace(-2.0, 6.0, 2000)
p1 = gaussian(t, 0.0, 0.7)
p2 = gaussian(t, 2.0, 0.7)
p3 = gaussian(t, 4.0, 0.7)

# Plot individual pulses (dashed) and their sum (solid)
ax1.plot(t, p1, linestyle="--", linewidth=2, label="symbol n-1")
ax1.plot(t, p2, linestyle="--", linewidth=2, label="symbol n")
ax1.plot(t, p3, linestyle="--", linewidth=2, label="symbol n+1")
ax1.plot(t, p1+p2+p3, linewidth=2.5, label="received (with ISI)")

ax1.set_title("ISI (time domain)")
ax1.set_xlabel("time")
ax1.set_ylabel("amplitude")
ax1.set_ylim(-0.05, 2.6)
ax1.legend(loc="upper right", frameon=False)

# Annotate overlap region
overlap_t = (t > 1.0) & (t < 3.0)
ax1.fill_between(t[overlap_t], p1[overlap_t]+p2[overlap_t], p1[overlap_t]+p2[overlap_t]+0.02, alpha=0.2)
ax1.text(1.4, 2.2, "overlap → ISI")

# ===== Right panel: Hofstadter-style lattice schematic =====
ax2 = fig.add_subplot(gs[0,1])

# Grid of lattice points
m_vals = np.arange(-3, 4)
n_vals = np.arange(-3, 4)
for m in m_vals:
    for n in n_vals:
        ax2.plot(m, n, marker="o", markersize=4)

# Highlight central site and nearest-neighbor hopping
center = (0,0)
neighbors = [(1,0), (-1,0), (0,1), (0,-1)]
for (dx, dy) in neighbors:
    ax2.arrow(center[0], center[1], dx*0.8, dy*0.8,
              head_width=0.12, head_length=0.15, length_includes_head=True, linewidth=2)

# Axis labels as lattice indices
ax2.set_xlim(-3.5, 3.5)
ax2.set_ylim(-3.5, 3.5)
ax2.set_aspect("equal", adjustable="box")
ax2.set_xticks([]); ax2.set_yticks([])
ax2.set_title("Hofstadter Hamiltonian (space domain)")

# Annotate Peierls phase on a vertical hop
ax2.text(0.15, 1.2, r"$e^{i 2\pi \alpha n}$", fontsize=12)

# Draw a thin square plaquette to hint at magnetic flux
square = np.array([[0,0],[1,0],[1,1],[0,1],[0,0]], dtype=float) - 0.5
ax2.plot(square[:,0], square[:,1], linestyle="--", linewidth=1)
ax2.text(0.1, 0.1, r"$\Phi$", fontsize=11)

# Save files
png_path = "/mnt/data/isi_hofstadter_diagram.png"
svg_path = "/mnt/data/isi_hofstadter_diagram.svg"
fig.savefig(png_path, bbox_inches="tight")
fig.savefig(svg_path, bbox_inches="tight")
png_path, svg_path
```
