```python
import numpy as np
import matplotlib.pyplot as plt

# Constants
e = 1.602e-19      # Charge of electron (C)
m_e = 9.109e-31    # Electron mass (kg)

def kinetic_inductance(n, width, thickness, m=m_e):
    """
    Calculate kinetic inductance per unit length.

    Parameters:
    - n: carrier density (m^-3)
    - width: wire width (meters)
    - thickness: wire thickness (meters)
    - m: effective mass of charge carrier (default: electron mass)

    Returns:
    - L_k: kinetic inductance per unit length (H/m)
    """
    A = width * thickness  # cross-sectional area
    L_k = m / (n * e**2 * A)
    return L_k

# Sample parameters
width = 1e-6       # 1 μm
thickness = 50e-9  # 50 nm
n_values = np.logspace(21, 29, 500)  # range of carrier densities

# Calculate L_k over n
L_k_values = kinetic_inductance(n_values, width, thickness)

# Plotting
plt.figure(figsize=(8, 5))
plt.loglog(n_values, L_k_values, label=f'{width*1e6:.0f} μm × {thickness*1e9:.0f} nm wire')
plt.xlabel("Carrier Density n (m⁻³)")
plt.ylabel("Kinetic Inductance per unit length (H/m)")
plt.title("Kinetic Inductance vs Carrier Density")
plt.grid(True, which="both", ls="--")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
#!pip install sympy
import sympy as sp
from sympy.physics.quantum import Operator, Commutator
from sympy.abc import x, p, m, h, lamda

# Define quantum operators
x_op = Operator('x')
p_op = Operator('p')

# Define operator-valued potential: V = λ x p
V_op = lamda * x_op * p_op

# Hamiltonian
H_op = p_op**2 / (2 * m) + V_op

# Define commutation rule: [x, p] = iħ
comm_rules = {
    Commutator(x_op, p_op): sp.I * h,
    Commutator(p_op, x_op): -sp.I * h
}

# Compute [x, H]
comm_xH = Commutator(x_op, H_op).doit().expand()
comm_xH = comm_xH.subs(comm_rules)

# Compute velocity operator: v = [x, H] / (iħ)
v_op = comm_xH / (sp.I * h)

print("Velocity operator v = [x, H] / (iħ):")
sp.pprint(v_op)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.linalg import expm

# Parameters
N = 50                  # Hilbert space dimension
hbar = 1.0              # Planck's constant
m = 1.0                 # mass
omega = 1.0             # oscillator freq
lam = 0.2               # coupling in V(x,p)
dt = 0.05               # time step
steps = 200             # number of time steps

# Creation and annihilation operators
a = np.diag(np.sqrt(np.arange(1, N)), 1)
adag = np.diag(np.sqrt(np.arange(1, N)), -1)

# Position and momentum operators in harmonic oscillator units
x_op = (a + adag) * np.sqrt(hbar / (2 * m * omega))
p_op = 1j * (adag - a) * np.sqrt(hbar * m * omega / 2)

# Kinetic energy term
H_kin = (p_op @ p_op) / (2 * m)

# Operator-valued potential: V = λ x p (symmetrized)
V_op = lam * 0.5 * (x_op @ p_op + p_op @ x_op)

# Full Hamiltonian
H = H_kin + V_op

# Diagonalize H for time evolution operator
U_dt = expm(-1j * H * dt / hbar)

# Initial state: ground state
psi = np.zeros((N, 2), dtype=complex)
psi[0, 0] = 1.0

# Observables
x_expect, p_expect = [], []

for _ in range(steps):
    x_expect.append(np.real((psi.conj().T @ x_op @ psi)[0, 0]))
    p_expect.append(np.real((psi.conj().T @ p_op @ psi)[0, 0]))
    psi = U_dt @ psi  # evolve state

x_var = []

for _ in range(steps):
    x_mean = (psi.conj().T @ x_op @ psi)[0, 0]
    x2_mean = (psi.conj().T @ (x_op @ x_op) @ psi)[0, 0]
    x_var.append(np.real(x2_mean - x_mean**2))
    psi = U_dt @ psi


# Plotting
time = np.arange(steps) * dt
plt.figure(figsize=(6, 4))
plt.plot(time, x_var)
plt.title("Wavepacket Spreading: $\sigma_x^2(t)$")
plt.xlabel("Time")
plt.ylabel("Position Variance")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.linalg import expm, eigh

# Simulation parameters
N = 10                 # Hilbert space dimension per oscillator
hbar = 1.0             # Planck's constant
m = 1.0                # mass
omega = 1.0            # oscillator frequency
lam = 0.1              # coupling strength
dt = 0.05              # time step
steps = 100            # number of time steps

# Creation and annihilation operators for one mode
def create_annihilation_operators(N):
    a = np.diag(np.sqrt(np.arange(1, N)), 1)
    adag = np.diag(np.sqrt(np.arange(1, N)), -1)
    return a, adag

a1, adag1 = create_annihilation_operators(N)
a2, adag2 = create_annihilation_operators(N)

# Tensor identity
I = np.eye(N)

# Operators for each mode
x1 = np.kron((a1 + adag1) * np.sqrt(hbar / (2 * m * omega)), I)
p1 = np.kron(1j * (adag1 - a1) * np.sqrt(hbar * m * omega / 2), I)

x2 = np.kron(I, (a2 + adag2) * np.sqrt(hbar / (2 * m * omega)))
p2 = np.kron(I, 1j * (adag2 - a2) * np.sqrt(hbar * m * omega / 2))

# Hamiltonian: H = p1^2/2m + p2^2/2m + 1/2 m omega^2 (x1^2 + x2^2) + lambda x1 p2
H_kin = (p1 @ p1 + p2 @ p2) / (2 * m)
H_pot = 0.5 * m * omega**2 * (x1 @ x1 + x2 @ x2)
H_int = lam * (x1 @ p2)
H = H_kin + H_pot + H_int

# Time evolution operator
U_dt = expm(-1j * H * dt / hbar)

# Initial state: product of ground states
psi0 = np.zeros((N, 1))
psi0[0, 0] = 1.0
psi = np.kron(psi0, psi0)  # tensor product

# Track entanglement entropy
def von_neumann_entropy(rho):
    eigs = np.linalg.eigvalsh(rho)
    eigs = eigs[eigs > 1e-12]
    return -np.sum(eigs * np.log(eigs))

entropy = []

for _ in range(steps):
    # Density matrix
    rho_full = psi @ psi.conj().T
    rho_full = rho_full.reshape(N, N, N, N)

    # Partial trace over mode 2
    rho_1 = np.einsum('ijlk->ik', rho_full)

    # Compute entropy
    S = von_neumann_entropy(rho_1)
    entropy.append(S)

    # Time evolve
    psi = U_dt @ psi

# Plot entropy
time = np.arange(steps) * dt
plt.figure(figsize=(8, 5))
plt.plot(time, entropy, label='Entanglement Entropy $S_1(t)$')
plt.xlabel("Time")
plt.ylabel("Entropy")
plt.title("Entanglement Entropy Growth due to $x_1 p_2$ Coupling")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import simpson
import pandas as pd

# Define time domain
t0, t1 = 0, 10
N = 500
t = np.linspace(t0, t1, N)
dt = t[1] - t[0]

# Define a trial trajectory: q(t) = sin(t)

oscillations = np.sin(1 * np.pi * t)  # fast oscillation
envelope = 319 * (1 - np.abs(t - 1.43) / 1.43)  # symmetric envelope
q = oscillations
#q = envelope * oscillations  # modulated oscillations

#q = np.sin(t)

# Define a small variation: delta_q(t) = epsilon * sin(pi * (t - t0)/(t1 - t0))
epsilon = 0.1
delta_q = epsilon * np.sin(np.pi * (t - t0) / (t1 - t0))

# Perturbed path
q_perturbed = q + delta_q

# Lagrangian: L = 0.5 * m * (dq/dt)^2 - V(q)
m = 1.0
V = lambda q: 0.5 * q**2  # Harmonic potential

def lagrangian(q, dqdt):
    return 0.5 * m * dqdt**2 - V(q)

# Compute derivatives
dqdt = np.gradient(q, dt)
dqdt_perturbed = np.gradient(q_perturbed, dt)

# Compute Lagrangians
L = lagrangian(q, dqdt)
L_perturbed = lagrangian(q_perturbed, dqdt_perturbed)

# Compute action via numerical integration (Simpson's rule)
S = simpson(L, t)
S_perturbed = simpson(L_perturbed, t)
delta_S = S_perturbed - S

# Plot the original and perturbed paths
plt.figure(figsize=(10, 5))
plt.plot(t, q, label='Original Path q(t)')
plt.plot(t, q_perturbed, label='Perturbed Path q(t) + δq(t)', linestyle='--')
plt.xlabel('Time t')
plt.ylabel('q(t)')
plt.title('Original vs Perturbed Trajectories')
plt.legend()
plt.grid(True)
plt.tight_layout()


# Create dataframe properly
action_df = pd.DataFrame({
    "Action": ["Original", "Perturbed", "Delta (S_perturbed - S)"],
    "Value": [S, S_perturbed, delta_S]
})
```
